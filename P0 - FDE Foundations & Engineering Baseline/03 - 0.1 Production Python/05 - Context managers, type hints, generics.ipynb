```python
    ############    #############   Context Managers, Type Hints, Generics   #############   ##############   

 =>  A context manager guarantees setup/teardown (acquire/release, open/close) runs even if
       an exception is raised in between -- 'with' is safer than manual try/finally scattered
       everywhere.

 =>  Type hints turn implicit assumptions ('this returns a list of users') into something a
       type checker (mypy/pyright) and your editor can verify.

 =>  Generics (TypeVar) let one function/class be correctly typed for many concrete types,
       instead of writing the same repository class once per model.
```

```python
from contextlib import contextmanager
from typing import Generic, TypeVar

@contextmanager
def timed_block(label: str):
    import time
    start = time.perf_counter()
    try:
        yield
    finally:
        print(f"{label}: {time.perf_counter() - start:.4f}s")

with timed_block("sleep-ish work"):
    sum(i * i for i in range(1_000_000))

T = TypeVar("T")

class Repository(Generic[T]):
    def __init__(self) -> None:
        self._items: list[T] = []

    def add(self, item: T) -> None:
        self._items.append(item)

    def all(self) -> list[T]:
        return list(self._items)

user_repo: Repository[str] = Repository()
user_repo.add("shriman@example.com")
print(user_repo.all())
```

```python
 =>  'Repository[str]' is the same class reused for any model type -- a type checker will
       flag 'user_repo.add(123)' as an error, catching a bug before it ever runs.

 =>  The same Repository[T] pattern is exactly what you'll reuse in Phase 3 for a real
       Postgres-backed repository layer.
```

```python
    ############    #############   Common Pitfalls   #############   ##############   

 =>  Putting code that can raise BEFORE the try in a manual context manager -- cleanup in
       the finally never runs if setup itself fails.

 =>  Adding type hints but never actually running a type checker in CI -- unchecked hints are
       just comments; they don't catch anything on their own.
```
