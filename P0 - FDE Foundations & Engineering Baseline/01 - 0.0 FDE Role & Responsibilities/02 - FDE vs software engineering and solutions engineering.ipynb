```python
    ############    #############   Why Bother Distinguishing These Roles?   #############   ##############   

 =>  Titles like 'Solutions Engineer', 'Sales Engineer', 'Forward Deployed Engineer',
       'Software Engineer', 'Consultant', and 'Customer Success' overlap in a lot of job
       postings -- but they answer to different goals, different time horizons, and
       different definitions of 'done'.

 =>  Knowing exactly which hat you're wearing on a given day tells you what 'good' looks
       like: a Sales Engineer's demo succeeding is not the same as an FDE's deployment
       succeeding.
```

| Role | Primary goal | Time horizon | Typical deliverable | Technical depth |
|---|---|---|---|---|
| **Sales / Solutions Engineer** | Win the deal -- prove technical feasibility | Days-weeks (pre-sale) | Demo, technical scoping doc | Medium -- broad, not deep |
| **Forward Deployed Engineer** | Ship a working solution inside the client's real environment | Weeks-months, embedded | Running, production-bound system | High -- must actually build it |
| **Software Engineer (product)** | Build/maintain the reusable core product | Ongoing, sprint-based | Shipped product features, stable platform | High -- deep in one codebase |
| **Consultant** | Advise on strategy/process; may not touch code | Weeks-months | Recommendations, roadmap, sometimes a PoC | Variable, often lower |
| **Customer Success** | Keep the customer adopted, renewed, expanding | Ongoing, post-sale | Adoption metrics, escalation handling | Low-medium |

```python
    ############    #############   The Core Distinctions   #############   ##############   

 =>  FDE vs Software Engineer: both write real production code, but a product SWE builds
       for thousands of unseen future customers through a roadmap process; an FDE builds for
       one real client in front of them, under that client's actual constraints, right now.
       (An FDE's code often *becomes* product code later -- but that's a side effect, not
       the immediate goal.)

 =>  FDE vs Solutions/Sales Engineer: a Solutions Engineer's job is done when the prospect
       is convinced the platform *can* solve their problem. An FDE's job isn't done until
       the thing is actually built, integrated, and running against real data -- 'it can
       work' vs 'it works, right now, in your environment.'

 =>  FDE vs Consultant: a consultant can hand over a report and leave. An FDE is judged on
       a running system, so a recommendation that isn't buildable in the client's actual
       environment isn't useful -- FDEs tend to prototype instead of just advising.

 =>  FDE vs Customer Success: CS optimizes an already-deployed relationship (renewal,
       expansion, support tickets). The FDE is typically most involved *before* that steady
       state exists -- getting the first real deployment to work at all.
```

```python
    ############    #############   The Handoffs (and Where They Go Wrong)   #############   ##############   

 =>  Sales/Solutions Engineer -> FDE: the handoff fails when the SE has over-promised
       something the FDE now has to somehow build in the client's actual (messier) reality.

 =>  FDE -> Software Engineer: the handoff fails when the FDE's field-built solution is too
       tightly coupled to one client's quirks to generalize -- or when nothing is documented
       and the SWE team can't maintain what they inherit.

 =>  FDE -> Customer Success: the handoff fails when the FDE leaves without transferring
       the operational knowledge (what's fragile, what was a workaround, what the client
       actually cares about) that CS now needs to keep the relationship healthy.

 =>  A good FDE actively manages both handoffs -- writes things down, flags what's a
       one-off hack vs. what's reusable, and doesn't just optimize for 'it demoed well.'
```

```python
    ############    #############   Common Pitfalls   #############   ##############   

 =>  Assuming the title on someone's badge tells you their actual function -- titles
       vary wildly by company; always check the 4 dimensions (goal, time horizon,
       deliverable, technical depth) instead of the label.

 =>  Letting one person silently absorb 3 roles at once (e.g. an FDE also doing full
       Sales Engineering and Customer Success) without naming that overload -- it hides
       a resourcing gap until something breaks.

 =>  Treating the handoffs as a one-time event instead of an ongoing relationship -- an
       FDE who 'throws code over the wall' to Software Engineering once and disappears
       leaves the next team guessing at every undocumented decision.
```
