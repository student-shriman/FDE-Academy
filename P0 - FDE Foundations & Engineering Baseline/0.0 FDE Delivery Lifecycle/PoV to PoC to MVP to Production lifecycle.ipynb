```python
    ############    #############   Four Stages, Four Different Questions   #############   ##############   

 =>  Every stage of this lifecycle exists to answer ONE question cheaply before you spend
       real money answering the next one. Skipping a stage (or not having an explicit exit
       criterion for it) is the #1 reason AI projects drag on for months with nothing shipped.

 =>  PoV (Proof of Value)  -- is this problem worth solving at all? No code required --
       sometimes a spreadsheet model of the expected ROI is enough.

 =>  PoC (Proof of Concept) -- can it technically be done? Thrown-away code is fine here --
       you're testing feasibility, not building anything that survives.

 =>  MVP (Minimum Viable Product) -- does a real, narrow version work against real users and
       real data, integrated into the client's actual systems?

 =>  Production -- is it reliable, monitored, and actually used every day, with an owner
       on-call for it like any other production system?
```

```python
    ############    #############   Exit Criteria -- the Part Everyone Skips   #############   ##############   

 =>  A stage without a written exit criterion never ends -- it just quietly turns into the
       next stage without anyone deciding that on purpose.

 =>  PoV exit: stakeholders (including whoever controls budget) agree, in writing, that the
       problem and its expected ROI are real -- before any engineering time is spent.

 =>  PoC exit: a narrow slice works against sample/synthetic data, in a notebook or a
       throwaway script -- proves the *idea* works, proves nothing about production-readiness.

 =>  MVP exit: real users are touching it, on real (not sample) data, integrated with the
       client's actual systems, and you can point to a KPI that moved.

 =>  Production exit: there's an on-call rotation, SLAs, monitoring, and a rollback plan --
       it's now just 'a system', not 'a project'.
```

```python
    ############    #############   The Most Common Failure Mode   #############   ##############   

 =>  Jumping straight from PoC to Production: a notebook that worked on 50 hand-picked
       examples gets wired directly into a client's live workflow, with no MVP stage to
       surface integration problems, edge cases, or real-usage failure modes first.

 =>  The opposite failure: getting stuck in PoC forever -- endlessly improving a demo that
       never touches real data or real users, because nobody defined what 'good enough to
       promote to MVP' actually means.

 =>  As an FDE, your job is often to notice which stage a project is *actually* in --
       regardless of what stage everyone believes it's in -- and get an explicit exit
       criterion agreed before moving on.
```
