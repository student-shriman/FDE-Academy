```python
    ############    #############   Why Map the Process Before Picking a Solution   #############   ##############   

 =>  Most 'AI solution' requests actually describe a symptom ("claims take too long"), not
       the underlying process. If you jump straight to a technical solution without mapping
       the real process, you risk automating the wrong step -- or automating a step that
       shouldn't exist at all.

 =>  A business-process map makes five things explicit: the people involved, the systems
       they touch, the data that flows between them, the dependencies/handoffs, and where
       it actually breaks down (the bottleneck).

 =>  You don't need special software for this -- lanes (one per person/system) and boxes
       (one per step) on a whiteboard, in Miro, or in matplotlib (as below) is enough.
```

```python
    ############    #############   Reading This Example   #############   ##############   

 =>  Four lanes: Customer, Intake Rep, CRM System, Claims Analyst -- one lane per distinct
       actor (human or system) in the process.

 =>  Two bottlenecks are visible the moment it's drawn out: manual re-entry (error-prone,
       no validation) and a batch/manual check each morning (up to 24h of pure waiting, with
       no automatic routing or priority).

 =>  Notice neither bottleneck is 'the analyst is slow' -- they're both process/system gaps.
       That's the point of mapping it: the map usually reveals that the fix is process
       automation (auto-routing, validation, structured intake) rather than 'make an AI
       model' -- or that AI is only one piece next to simpler fixes.
```

```python
    ############    #############   How to Actually Do This With a Client   #############   ##############   

 =>  Interview the people who actually do the work today, not just their manager -- the
       manager describes the process as designed; the front-line worker describes the
       workarounds they've built because the designed process doesn't quite work.

 =>  Ask 'what do you do right before this?' and 'what do you do right after?' repeatedly --
       this is how you discover undocumented manual steps (a spreadsheet nobody mentioned,
       a Slack message that's actually part of the approval chain).

 =>  Mark every point where data changes format or system (verbal -> CRM, CRM -> Excel,
       Excel -> email) -- these transitions are where errors, delay, and lost context
       concentrate.

 =>  Timestamp each step if you can get the data -- 'feels slow' is not the same as 'takes
       18 hours on average, with a long tail up to 4 days.'
```

```python
    ############    #############   Hands-on Lab Checklist   #############   ##############   

 =>  [ ] Pick a real (or realistic hypothetical) workflow you have some visibility into.
           Draw it as swimlanes: one lane per person/system involved.

 =>  [ ] Mark every manual data re-entry point and every place a human waits on another
           human or a batch/manual check.

 =>  [ ] For each bottleneck you find, write one sentence: is the fix 'automate this exact
           step', 'remove this step entirely', or 'this isn't actually the real bottleneck'?
```

```python
    ############    #############   Common Pitfalls   #############   ##############   

 =>  Mapping only the 'happy path' -- the exceptions, retries, and manual workarounds are
       usually where the real cost and risk live, not the case that always works cleanly.

 =>  Mapping the process as management believes it works, not as front-line staff
       actually run it today -- the gap between the two is often the most valuable
       finding of the whole exercise.

 =>  Jumping to 'automate this with AI' before finishing the map -- some bottlenecks
       (a missing validation rule, a step that shouldn't exist) are cheaper to fix with
       a process change than with any model.
```
