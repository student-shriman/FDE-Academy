```python
    ############    #############   Turning "Build Us a Chatbot" Into Something Buildable   #############   ##############   

 =>  An ambiguous request like 'build us a chatbot' isn't a requirement -- it's a starting
       point. Every word in it is hiding a decision someone still needs to make.

 =>  The FDE's job is to fan that one sentence out into concrete, answerable questions
       across five dimensions, then fan those answers back in to one thing: acceptance
       criteria you can actually build against and test.
```

```python
    ############    #############   The Five Questions, One at a Time   #############   ##############   

 =>  Users -- who, exactly? 'Employees' is not an answer -- is it the 40-person support team,
       or every employee in the company? What access/permissions do they have today?

 =>  Workflow -- what specific task, done today, does this replace or assist? If you can't
       name the current manual process, you don't have a real requirement yet.

 =>  Constraints -- what can't change? Data residency rules, a legacy system you're not
       allowed to touch, a latency ceiling, an approved model/vendor list, a budget.

 =>  Requirements -- of everything discussed, what's actually must-have for a first version
       vs. nice-to-have later? (This is where 'shippable increments' from the next notebook
       comes in.)

 =>  KPIs & risks -- how will you know, with a number, whether this worked? And what's the
       worst thing that happens if it's wrong (a bad answer shown to a customer vs. an
       internal draft a human reviews)?
```

```python
    ############    #############   A Worked Example   #############   ##############   

 =>  Request: "Build us a chatbot."

 =>  Users: the 12-person billing support team, internal use only, not customer-facing.

 =>  Workflow: today they manually search 3 systems (CRM, billing DB, a PDF policy doc) to
       answer a customer's billing question -- average 6 minutes per ticket.

 =>  Constraints: billing data cannot leave the company's Azure tenant; no vendor outside
       the already-approved model list; must integrate with existing ticketing tool.

 =>  Requirements (v1, must-have): answer from the PDF policy doc + billing DB read access;
       cite its source; escalate to a human when unsure. (Nice-to-have, later: write access
       to update tickets automatically.)

 =>  KPIs & risks: reduce average handling time from 6 min to under 2 min; risk = a wrong
       answer about a refund amount, so every AI answer is shown to the agent, not the
       customer, until accuracy is proven over a pilot period.

 =>  Notice how different this is from "build us a chatbot" -- and how much of it was not
       in the original sentence at all.
```
