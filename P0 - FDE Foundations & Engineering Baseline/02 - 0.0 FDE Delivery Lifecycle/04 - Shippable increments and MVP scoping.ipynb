```python
    ############    #############   Slice Vertically, Not Horizontally   #############   ##############   

 =>  A tempting way to plan an MVP is layer-by-layer: get the UI to 80%, the API to 80%,
       the business logic to 80%, the data layer to 80% -- and it feels like steady progress
       right up until the point where NOTHING works end-to-end, because every layer is
       still missing the other 20%.

 =>  The alternative: build one thin vertical slice -- a single, narrow path through every
       layer (UI -> API -> logic -> data) that is fully, genuinely done -- before adding a
       second slice.

 =>  This is sometimes called a 'walking skeleton': not a full-featured app, but a complete,
       if narrow, real system that already runs end-to-end in the client's environment.
```

```python
    ############    #############   Why This Matters More for an FDE Than a Product Team   #############   ##############   

 =>  A product team can often absorb a few months of 'nothing demos yet' because they're
       building for a roadmap. An FDE is usually accountable to one client watching the
       clock -- and 'we're 80% done with everything' is much harder to defend than 'here is
       one real thing, working, today.'

 =>  Vertical slices also surface integration problems (auth, the client's real data
       quality, a legacy system's quirks) on day one instead of in week 10 -- exactly the
       kind of surprise that turns an MVP deadline into a crisis.

 =>  Each slice should be chosen to reduce the biggest remaining risk or uncertainty first --
       not the easiest slice, the one that answers the scariest open question.
```

```python
    ############    #############   Scoping an MVP: A Practical Method   #############   ##############   

 =>  1. List every capability someone has asked for or implied.

 =>  2. For each one, ask: does leaving this out of v1 make the whole thing useless, or just
           less convenient? Useless -> must-have slice #1. Less convenient -> later slice.

 =>  3. Order the must-have slices by risk: the slice most likely to reveal a nasty surprise
           (an unfamiliar legacy system, an ambiguous data source, an unclear permission
           model) goes first, even if it's not the most visually impressive.

 =>  4. Write down explicitly what is deliberately NOT in v1 -- an unscoped 'nice-to-have'
           silently becomes scope creep the moment nobody wrote down that it was cut.
```

```python
    ############    #############   Common Pitfalls   #############   ##############   

 =>  Picking the easiest slice first instead of the riskiest -- shipping something
       impressive-but-safe first just delays discovering the hard problem (an unfamiliar
       legacy system, an ambiguous permission model) until later, when it's more expensive
       to fix.

 =>  Not treating each slice as a small experiment with its own dependencies -- a slice
       that silently depends on another team's unfinished work isn't actually shippable,
       even if your own code is done.

 =>  Never writing down what was deliberately cut from v1 -- an unrecorded 'nice-to-have'
       quietly turns into an assumed requirement the moment someone notices it's missing.
```
