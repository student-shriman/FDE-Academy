```python
    ############    #############   Trunk-Based Development   #############   ##############   

 =>  main is always deployable. Every change is a short-lived branch (hours to a few days,
       not weeks) that merges back via a small, reviewed pull request.

 =>  Short branches exist because merge conflict difficulty grows roughly with how long a
       branch has diverged AND how many other people also branched off the same point --
       merging daily keeps that cost small and predictable.
```

```python
    ############    #############   Git Commands You'll Actually Use Daily   #############   ##############   

      git checkout -b feature/add-login      # branch off main
      git add -p                             # stage hunks selectively, not whole files
      git commit -m "..."                    # small, focused commits
      git fetch origin && git rebase origin/main   # stay current with main, avoid drift
      git push -u origin feature/add-login
      gh pr create                           # open a small, reviewable PR

 =>  'git add -p' matters more than it looks -- it forces you to actually look at every
       hunk before committing, catching stray debug prints/commented-out code before they
       ship.
```

```python
    ############    #############   What Makes a Pull Request Actually Reviewable   #############   ##############   

 =>  Small: a 50-line PR gets a careful review; a 2000-line PR gets a rubber stamp -- size
       is the single biggest lever you control over review quality.

 =>  One logical change per PR: don't mix a refactor with a new feature -- a reviewer can't
       tell which lines are 'behavior change' vs 'just moved code' at the same time.

 =>  A description that answers 'why', not just 'what' -- the diff already shows what
       changed; the description should explain the reasoning a diff can't show.

 =>  Self-review first: read your own diff before requesting review -- most obvious issues
       (leftover prints, a forgotten TODO) are caught faster by the author than by a
       reviewer.
```

```python
    ############    #############   Hands-on Lab Checklist   #############   ##############   

 =>  [ ] Practice 'git add -p' on a real change with 2+ unrelated edits, and split it into
           2 focused commits.

 =>  [ ] Write an ADR-style PR description (context, decision, why, alternatives
           considered) for your next real change, even a small one.
```

```python
    ############    #############   Common Pitfalls   #############   ##############   

 =>  Long-lived branches that drift for weeks -- the eventual merge becomes a high-risk
       event instead of a routine one.

 =>  Reviewing a PR by only reading the diff, never checking it out and running it -- some
       bugs (a broken import, a missing migration) only show up when the code actually runs.

 =>  Force-pushing over a branch others are also reviewing/building on without warning --
       rewrites their local history and confuses in-progress review comments.
```
